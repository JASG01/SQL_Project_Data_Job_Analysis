```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from numpy import savetxt
```

```python
df=pd.read_csv('2_top_paying_job_skills.csv')
df
```

```python
skill_counts = df['skills'].value_counts()

# Plot the horizontal bar chart
# Calling .plot.barh() directly on the Series
top_10_skills = df['skills'].value_counts().head(10)

# Step 3: Plot the top 10 skills as a horizontal bar chart
ax = top_10_skills.plot.barh(figsize=(10, 6))

# Step 4: Customize and display the plot
ax.set_title('Top 10 Skills in High-Paying Data Analyst Roles (2023)', fontsize =18)
ax.set_xlabel('Frequency', fontsize =16)
ax.set_ylabel('Skill', fontsize =16)

# Optional: Invert the y-axis so the highest count is at the top
ax.invert_yaxis()

plt.tight_layout()
plt.savefig("skills_bar_plot.png", facecolor='white', dpi=500)
plt.show()
```
